1.  Write a Python function to find the maximum of two numbers.

```python
def max_numbers(numbers):
    
    max_of_two = max(numbers[0], numbers[1])

    return max_of_two

# Example usage
numbers = [1, 8]
print(max_numbers(numbers)) # Output: 8
```

2.  Write a Python function to find the minimum of two numbers.

```python
def min_numbers(numbers):
    
    min_of_two = min(numbers[0], numbers[1])

    return min_of_two

# Example usage
numbers = [1, 8]
print(min_numbers(numbers)) # Output: 1
```

3.  Write a Python function to check if a given number is even or odd.

```python
def odd_even_numbers(number):
    if number % 2 == 0:
        return "Even"
    else:
        return "Odd"

# Example usage
print(odd_even_numbers(1))   
print(odd_even_numbers(2))   
print(odd_even_numbers(3))
```

4.  Write a Python program to print all the even numbers from 1 to 100.

```python
def one_to_hundred_even():
    for number in range(0, 101, 2):
        print(number)

# Example usage
one_to_hundred_even()# Output: 0
```

5.  Write a Python program to print all the odd numbers from 1 to 100.

```python
def one_to_hundred_odd():
    for number in range(1, 101, 2):
        print(number)

# Example usage
one_to_hundred_odd()# Output: 1
```

6.  Write a Python program to find the sum of all the numbers from 1 to 100.

```python
def one_to_hundred_sum():
    total_sum = sum(range(1, 101))
    print(total_sum)

# Example usage
one_to_hundred_sum()# Output: 5050
```

7.  Write a Python program to find the factorial of a given number.

```python
def get_factorial_recursive(n):
    if n < 0:
        raise ValueError("dumb dumb you should know that factorials are only defined for non-negative integers")
    if n == 0 or n == 1:
        return 1
    
    return n * get_factorial_recursive(n - 1)

print(get_factorial_recursive(4))  
```
